```python
x = new_datasets.drop(columns='Class',axis=1)
y = new_datasets['Class']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
li = LinearRegression()
```

```python
li.fit(x_train,y_train)
```

```python
x_train_prediction = li.predict(x_train)

x_train_prediction_rounded = x_train_prediction.round().astype(int)

datasets_accuracy = accuracy_score(x_train_prediction_rounded, y_train)

print(f"Accuracy (rounded): {datasets_accuracy}")
```
